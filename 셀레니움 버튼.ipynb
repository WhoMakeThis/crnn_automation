```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
```

```python
from selenium.webdriver.common.by import By
```

```python
# 브라우저 실행 파일 경로 설정
chrome_binary_path = "C://Users//user//Downloads//chrome-win64//chrome-win64//chrome.exe"

# 크롬 드라이버 경로 설정
chrome_driver_path = "C://Users//user//Downloads//chromedriver-win64//chromedriver-win64//chromedriver.exe"  # 크롬 드라이버의 실제 경로로 수정

options = Options()
options.binary_location = chrome_binary_path

service = Service(executable_path=chrome_driver_path)
```

```python
# 브라우저 실행
browser = webdriver.Chrome(service=service, options=options)
# start_url = 'https://www.youtube.com'
start_url = 'http://211.188.49.36:5000/'
browser.get(start_url)
browser.maximize_window()
```

```python
# 요소 찾고 클릭하기
element = browser.find_elements(By.XPATH, '/html/body/div/div[1]/button[1]')[0]
element.click()
```
